En electrónica, a menudo necesitamos manejar diferentes tipos de datos: lecturas de sensores, estados de dispositivos, configuraciones fijas, etc. Python ofrece varias estructuras de datos para organizar esta información eficientemente.

1. Tuplas

Las tuplas son estructuras inmutables, es decir, no se pueden modificar. Útiles para datos fijos como pines de conexión o rangos de calibración.

```python
motor_pasos = (17, 18, 27, 22)  # GPIO pins
for pin in motor_pasos:
    print(f"Activando pin GPIO {pin}")
```

2. listas 

Las listas son estructuras mutables que permiten agregar, modificar o eliminar elementos. Son ideales para almacenar lecturas de sensores o series de datos.

```python
#sensor DHT11
from random import uniform
lecturas = [round(uniform(20.0, 30.0), 1) for _ in range(5)]  # Simulando sensor DHT11
promedio = sum(lecturas) / len(lecturas)
print(f"Lecturas: {lecturas} | Promedio: {promedio:.1f}°C")
```

3. diccionarios

Los diccionarios almacenan pares clave-valor, ideales para representar múltiples sensores con sus datos.

```python
#estados de conectividad "leds"
dispositivos = {
    "led_rojo": False,
    "ventilador": True,
    "buzzer": False
}
for nombre, estado in dispositivos.items():
    print(f"{nombre.upper()} encendido: {'Sí' if estado else 'No'}")
```

4. conjuntos 

Los conjuntos eliminan automáticamente los valores duplicados, útiles para limpiar datos repetidos o detectar elementos únicos.

```python
gpio_total = set(range(2, 28))           # GPIO disponibles en Pi
ocupados = {17, 18, 22, 23}              # En uso
disponibles = gpio_total - ocupados
print("GPIO libres:", sorted(disponibles))
```
