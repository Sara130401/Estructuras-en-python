1. Tuplas (Datos Inmutables)

Ejercicio 1: Un sensor de presión tiene coeficientes de calibración definidos como una tupla (a, b, c). Crea una tupla con valores hipotéticos y usa la ecuación P = a*x^2 + b*x + c para calcular la presión con un valor de entrada x = 5.

```python
# Tupla de coeficientes
coeficientes = (0.8, -2.5, 10)

# Entrada del sensor
x = 5

# Cálculo de la presión
a, b, c = coeficientes
P = a * x**2 + b * x + c

print(f"Presión calculada: {P:.2f} unidades")
```

Ejercicio 2: Dada una tupla con las configuraciones de un ADC ((resolución_bits, voltaje_referencia)), extrae los valores individuales y calcula el valor de un LSB (Least Significant Bit).

```python
# Tupla con configuración del ADC
adc_config = (10, 3.3)  # (bits, voltaje de referencia)

# Extraer los valores
resolucion, vref = adc_config

# Calcular el valor del LSB
lsb = vref / (2**resolucion)
print(f"Valor del LSB: {lsb:.6f} V")
```

2. Listas (Datos Dinámicos)

Ejercicio 3: Simula la adquisición de 10 muestras de temperatura de un sensor, almacénalas en una lista y calcula el promedio de temperatura.

```python
import random

# Simulación de 10 lecturas de temperatura
temperaturas = [round(random.uniform(20.0, 30.0), 2) for _ in range(10)]

# Cálculo del promedio
promedio = sum(temperaturas) / len(temperaturas)

# Resultados
print("Lecturas:", temperaturas)
print(f"Promedio de temperatura: {promedio:.2f} °C")
```

Ejercicio 4: Un sensor de corriente ha registrado valores anómalos. Filtra una lista de mediciones eliminando los valores fuera del rango 10 ≤ I ≤ 50.

```python
# Lista con lecturas simuladas 
lecturas_corriente = [8, 12, 45, 55, 30, 7, 50, 11, 65, 22]

# Filtrado: mantener solo los valores entre 10 y 50 inclusive
lecturas_filtradas = [i for i in lecturas_corriente if 10 <= i <= 50]

# Resultados
print("Lecturas originales:", lecturas_corriente)
print("Lecturas válidas (10 ≤ I ≤ 50):", lecturas_filtradas)
```

3. Diccionarios (Asociación de Datos)

Ejercicio 5: Crea un diccionario para almacenar información de tres sensores (nombre, ubicación, unidad_medida). Luego, imprime los detalles de cada sensor.

```python
# Diccionario con información de tres sensores
sensores = {
    "sensor_1": {
        "nombre": "Temperatura",
        "ubicacion": "Sala de control",
        "unidad_medida": "°C"
    },
    "sensor_2": {
        "nombre": "Presión",
        "ubicacion": "Tubería principal",
        "unidad_medida": "psi"
    },
    "sensor_3": {
        "nombre": "Humedad",
        "ubicacion": "Almacén",
        "unidad_medida": "%"
    }
}

# Imprimir detalles de cada sensor
for clave, datos in sensores.items():
    print(f"\nID: {clave}")
    for campo, valor in datos.items():
        print(f"{campo.capitalize()}: {valor}")
```

Ejercicio 6: Dado un diccionario con lecturas de temperatura en diferentes momentos {"10:00 AM": 22.5, "10:10 AM": 23.1, "10:20 AM": 22.8}, agrega una nueva lectura y encuentra la temperatura más alta registrada.

```python
# Diccionario con lecturas horarias de temperatura
temperaturas = {
    "10:00 AM": 22.5,
    "10:10 AM": 23.1,
    "10:20 AM": 22.8
}

# Agregar una nueva lectura
temperaturas["10:30 AM"] = 23.6

# Encontrar la temperatura más alta
max_temp = max(temperaturas.values())

# Mostrar resultados
print("Lecturas registradas:")
for hora, temp in temperaturas.items():
    print(f"{hora} → {temp} °C")

print(f"\nTemperatura más alta: {max_temp} °C")
```

4. Conjuntos (Eliminación de Duplicados)

Ejercicio 7: Un sensor ha detectado eventos en un sistema eléctrico, pero algunos están duplicados. Usa un conjunto para mantener solo eventos únicos

```python
# Lista con eventos detectados (algunos repetidos)
eventos_detectados = ["sobrecarga", "baja_tension", "pico_corriente", "sobrecarga", "fallo_sensor", "pico_corriente"]

# Convertir la lista a un conjunto para eliminar duplicados
eventos_unicos = set(eventos_detectados)

# Mostrar resultados
print("Eventos originales:", eventos_detectados)
print("Eventos únicos detectados:", eventos_unicos)
```

Ejercicio 8: Dados dos conjuntos con eventos detectados en dos sensores diferentes, encuentra qué eventos ocurrieron en ambos sensores.

```python
# Eventos detectados por cada sensor
sensor_A_eventos = {"sobrecarga", "pico_corriente", "fallo_sensor", "rearme"}
sensor_B_eventos = {"pico_corriente", "sobrecarga", "baja_tension", "desconexion"}

# Encontrar eventos comunes (intersección)
eventos_comunes = sensor_A_eventos & sensor_B_eventos  # o sensor_A_eventos.intersection(sensor_B_eventos)

# Mostrar resultados
print("Eventos comunes en ambos sensores:", eventos_comunes)
```
