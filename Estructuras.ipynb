1. listas (list)

Cuando necesitas almacenar elementos modificables y ordenados.

Puedes añadir, eliminar o modificar elementos.

Se permiten elementos repetidos.

```python
frutas = ["manzana", "plátano", "naranja"]
frutas.append("sandía")
print(frutas)  # ['manzana', 'plátano', 'naranja', 'sandía']
```

2. tuplas (tuple)

Cuando necesitas una colección inmutable (no cambia).

Útil para representar datos fijos, como coordenadas o configuraciones.

```python
coordenada = (10.5, 20.3)
print(coordenada[0])  # 10.5
```

3. diccionarios (dict)

Cuando necesitas asociar claves con valores (tipo JSON).

Ideal para representar objetos o estructuras con propiedades.

```python
persona = {"nombre": "Ana", "edad": 25, "ciudad": "CDMX"}
persona["edad"] = 26  # Modificar valor
persona["profesión"] = "Ingeniera"  # Añadir nueva clave
print(persona)
```

4. conjuntos (set)

Cuando necesitas almacenar elementos únicos y sin orden.

Ideal para eliminar duplicados o hacer operaciones matemáticas como unión o intersección.

```python
colores = {"rojo", "verde", "azul"}
colores.add("amarillo")
colores.add("rojo")  # No se repite
print(colores)  # {'verde', 'rojo', 'azul', 'amarillo'}
```
